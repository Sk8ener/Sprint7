### obtenir_dades()
La función hará lo siguiente:
1. Recibirá opcionalmente una lista de datos de prueba.
2. Si recibe datos de prueba, obtendrá de la lista los seis datos del candidato.
3. Si no recibe una lista, solicitará los datos mediante input().
4. Devolverá una tupla con los seis datos en el orden esperado por     validar_dades().

```python

def obtenir_dades(inputs_test=None):
      ''' Recibe la lista otorgada para hacer pruebas, y si no la recibe
      continúa con el formulario para introducir los datos.'''

    # Código para obtener los datos de la lista de prueba

      if inputs_test is not None:
        nombre = inputs_test.pop(0)
        telefono = inputs_test.pop(0)
        entrada_anios = inputs_test.pop(0)
        entrada_catalan = inputs_test.pop(0)
        entrada_excel = inputs_test.pop(0)
        entrada_titulo = inputs_test.pop(0)

  # código para obtener los datos manualmente de los candidatos
      else:
        nombre = input("Introduce nombre y apellidos: ").strip().title()
        telefono = input("Introduce el teléfono: ").strip()
        entrada_anios = input("Introduce los años de experiencia: ").strip()

        print("Catalán: 0 = sin conocimiento, 1 = hasta B1, 2 = B2, 3 = C1 o superior")
        entrada_catalan = input("Puntuación de catalán (0-3): ").strip()

        print("Excel: 0 = básico, 1 = funciones, 2 = tablas dinámicas, 3 = Visual Basic y macros")
        entrada_excel = input("Puntuación de Excel (0-3): ").strip()

        print("Titulación: 0 = ESO, 1 = FP, 2 = Universidad, 3 = Máster o superior")
        entrada_titulo = input("Puntuación de titulación (0-3): ").strip()

      return (nombre, telefono, entrada_anios, entrada_catalan, entrada_excel, entrada_titulo)
```

### validar_dades()

1. Recibirá la tupla con los datos de obtenir_dades().
2. Comprobará el nombre y el teléfono.
3. Comprobará que los años de experiencia sean un número entre 0 y 40.
4. Comprobará que las puntuaciones de catalán, Excel y titulación sean enteros entre 0 y 3.
5. Si encuentra un error, mostrará un mensaje y devolverá None.
6. Si todos los datos son válidos, devolverá la tupla con los valores numéricos convertidos a sus tipos correspondientes.

```python

def validar_dades(datos):
    ''' Recibe la tupla de obtenir_dades(), comprueba que los datos sean correctos
    ya sea de la lista de pruebas o del formulario'''

    nombre = datos[0]
    telefono = datos[1]
    entrada_anios = datos[2]
    entrada_catalan = datos[3]
    entrada_excel = datos[4]
    entrada_titulo = datos[5]

    # Validar nombre
    if not nombre or not any(letra.isalpha() for letra in nombre):
        print("Inscripción rechazada: nombre no válido.")
        print("")
        return None

    # Validar teléfono
    if telefono.startswith("+"):
        numero = telefono[1:]
    else:
        numero = telefono

    if not numero.isdigit():
        print("Inscripción rechazada: teléfono no válido.")
        print("")
        return None

    # Validar años de experiencia
    try:
        anios = float(entrada_anios)
    except (ValueError, TypeError):
        print("Inscripción rechazada: años de experiencia no válidos.")
        print("")
        return None

    if anios < 0 or anios > 40:
        print("Inscripción rechazada: los años deben estar entre 0 y 40.")
        print("")
        return None

    # Validar catalán, Excel y titulación
    try:
        catalan = int(entrada_catalan)
        excel = int(entrada_excel)
        titulo = int(entrada_titulo)
    except (ValueError, TypeError):
        print("Inscripción rechazada: las puntuaciones deben ser enteros.")
        print("")
        return None

    if (catalan != float(entrada_catalan)
            or excel != float(entrada_excel)
            or titulo != float(entrada_titulo)):
        print("Inscripción rechazada: las puntuaciones deben ser enteros.")
        print("")
        return None

    if not (0 <= catalan <= 3
            and 0 <= excel <= 3
            and 0 <= titulo <= 3):
        print("Inscripción rechazada: las puntuaciones deben estar entre 0 y 3.")
        print("")
        return None

    # Todos los datos son válidos
    return (nombre, telefono, anios, catalan, excel, titulo)
```

### La función calcular_puntuacio() recibe los datos validados de un candidato y calcula su puntuación total según los criterios establecidos en el ejercicio.


1. Recibe la tupla con los datos del candidato.
2. Obtiene los años de experiencia, la puntuación de catalán, la puntuación de Excel y la puntuación de la titulación académica.
3. Multiplica los años de experiencia por 1,5.
4. Suma los puntos obtenidos en catalán, Excel y titulación.
5. Devuelve la puntuación total calculada.
La fórmula que utiliza es:
P=(A X 1.5) + C+ X + T

Donde:
- \(A\): años de experiencia en la empresa.
- \(C\): puntuación de catalán.
- \(X\): puntuación de Excel.
- \(T\): puntuación de la titulación académica.



Como validar_dades() devuelve los datos en este orden:

(nombre, telefono, anios, catalan, excel, titulo)
Se accederá a éstos por su índice.
Por ejemplo: datos[2] correspondiendo a anios.


```python

def calcular_puntuacio(datos):
    ''' Recibe la tupla con los datos del candidato y calcula su puntuación total.'''

    anios = datos[2]
    catalan = datos[3]
    excel = datos[4]
    titulo = datos[5]

    puntuacion = (anios * 1.5) + catalan + excel + titulo

    return puntuacion
```

### La función classificar_candidat() recibe la puntuación total de un candidato y le asigna una categoría según los criterios del ejercicio.
1. Recibe la variable puntuacion.
2. Comprueba si la puntuación es igual o superior a 12. Si lo es, devuelve "Prioridad alta".
3. Si no llega a 12, comprueba si es igual o superior a 8. Si lo es, devuelve "Candidato adecuado".
4. Si es inferior a 8, devuelve "No prioritario".

```python
def classificar_candidat(puntuacion):
    ''' Recibe la puntuación total del candidato y devuelve su clasificación.'''

    if puntuacion >= 12:
        return "Prioridad alta"
    elif puntuacion >= 8:
        return "Candidato adecuado"
    else:
        return "No prioritario"
```



### La función mostrar_individual() muestra por pantalla la información de un candidato, su puntuación total y su clasificación.

### La función mostrar_informe()

1. Recibe la lista candidatos, con los candidatos válidos.
2. Crea una nueva lista nueva llamada candidatos_ordenados.
3. Separa los candidadtos según su clasificación, respetando el orden de prioridad.
4. Ordena por puntuación, de mayor a menor, dentro de cada categoría.
5. Muestra el informe y devuelve la lista ordenada.

```python

def mostrar_informe(candidatos):
    ''' Recibe la lista de candidatos válidos. Crea una lista nueva ordenada. Se
    ordena la puntuación de mayor a menor, dentro de cada categoría. Muestra el
    informe y devuelve la lista ordenada'''

    candidatos_ordenados = candidatos.copy()

    for i in range(len(candidatos_ordenados)):
        for j in range(i + 1, len(candidatos_ordenados)):
            prioridad_i = candidatos_ordenados[i][3]
            prioridad_j = candidatos_ordenados[j][3]

            if prioridad_i == "Prioridad alta":
                nivel_i = 3
            elif prioridad_i == "Candidato adecuado":
                nivel_i = 2
            else:
                nivel_i = 1

            if prioridad_j == "Prioridad alta":
                nivel_j = 3
            elif prioridad_j == "Candidato adecuado":
                nivel_j = 2
            else:
                nivel_j = 1

            if nivel_j > nivel_i or (
                nivel_j == nivel_i
                and candidatos_ordenados[j][2] > candidatos_ordenados[i][2]):
                temporal = candidatos_ordenados[i]
                candidatos_ordenados[i] = candidatos_ordenados[j]
                candidatos_ordenados[j] = temporal

    for candidato in candidatos_ordenados:
        print("Nombre:", candidato[0])
        print("Teléfono:", candidato[1])
        print("Puntuación:", candidato[2])
        print("Clasificación:", candidato[3])
        print("")

    return candidatos_ordenados
```

###  Algoritmo de main_seleccio()

1. Crear una lista vacía llamada candidatos.
2. Mientras haya menos de 10 candidatos aceptados:
   - Recoger los datos con obtenir_dades().
   - Validarlos con validar_dades().
   - Si los datos son incorrectos, rechazar la inscripción y pasar al siguiente candidato.
   - Si son correctos, calcular la puntuación con calcular_puntuacio().
   - Clasificar al candidato con classificar_candidat().
   - Guardar sus datos, puntuación y clasificación en la lista.
   - Mostrar su resultado individual con mostrar_individual().
3. Cuando haya 10 candidatos válidos, generar el informe final con mostrar_informe().

```python

def main_seleccio(inputs_test=None):
    ''' Crea una lista Vacía, recibe hasta 10  con obetenir_dades(), los valida
    con validar_dades() y los agrega sin son correctos, si no lo son los rechaza,
     después llamava las funciones calcular_puntuacio(), classificar_candidat(), muestra el resultado
     individual con mostrar_individual() y cuando hay 10 candidatos Válidos genera
     el informe final con mostrar_informe() '''

    candidatos = []

    while len(candidatos) < 10:
        datos = obtenir_dades(inputs_test)
        datos_validados = validar_dades(datos)

        if datos_validados is None:
            print("Inscripción rechazada. Pasamos al siguiente candidato.")
            print("")

        else:
            puntuacion = calcular_puntuacio(datos_validados)
            clasificacion = classificar_candidat(puntuacion)

            candidato = (
                datos_validados[0],
                datos_validados[1],
                puntuacion,
                clasificacion )

            candidatos.append(candidato)

            mostrar_individual(datos_validados,puntuacion,clasificacion )
    print("Inscripción finalizada. Generando informe...")
    print("")
    print("Informe final:")
    print("")
    mostrar_informe(candidatos)
```

#### obtenir_segment_i_factor()

Esta función recibe un código de comportamiento y devuelve un diccionario con el segmento, el mensaje y el factor correspondientes.

Para ello se crea un diccionario de diccionarios: cada código es una clave exterior y su valor contiene tres datos.

Cómo funciona

Si la función recibe el valor 4, busca esa clave y devuelve el diccionario correspondiente. De ese diccionario podemos recuperar el segmento, el mensaje y el factor 0.90.

Nota: Los mensajes son adecuados al tema para motivar al cliente a que realice la compra, porque el enunciado del ejercicio no proporciona textos concretos.

```python

def obtenir_segment_i_factor(codi_comportament):
    ''' Recibe un código de comportamiento y devuelve un diccionario con el segmento,
    el mensaje y el factor correspondientes.'''

    segmento_factor = {}

    segmento_factor[0] = {
        "segmento": "Navega pero no añade productos",
        "mensaje": "Descubre productos que te pueden interesar",
        "factor": 0.10}

    segmento_factor[1] = {
        "segmento": "Añade productos pero no llega al checkout",
        "mensaje": "Tu cesta te está esperando",
        "factor": 0.25}

    segmento_factor[2] = {
        "segmento": "Llega al checkout pero no paga",
        "mensaje": "Completa tu compra de forma sencilla",
        "factor": 0.50}

    segmento_factor[3] = {
        "segmento": "Abandono de cesta hace menos de 24 h",
        "mensaje": "No olvides terminar tu compra",
        "factor": 0.75}

    segmento_factor[4] = {
        "segmento": "Abandono de cesta hace más de 24 h",
        "mensaje": "Vuelve a tu cesta y descubre lo que te espera",
        "factor": 0.90}

    return segmento_factor[codi_comportament]
```

#### obtenir_factor_edat()

Esta función recibe el grupo de edad del cliente y devuelve el factor numérico asignado a ese grupo. No necesita conocer el resto de los datos del cliente.

Cómo funciona

Si recibe "18-25", consulta el diccionario y devuelve 1.20.

Utilizamos un diccionario porque cada grupo de edad ya tiene asociado un factor concreto.

```python

def obtenir_factor_edat(grup_edat):
    ''' Recibe el grupo de edad del cliente y devuelve el factor numérico
    asignado a ese grupo.'''
    factores_edad = {
        "18-25": 1.20,
        "26-40": 1.00,
        "41-60": 0.85,
        "60+": 0.70}

    return factores_edad[grup_edat]
```

#### calcular_risc()

Esta función recibe los dos factores obtenidos anteriormente, calcula su producto y devuelve el riesgo final.

Cómo funciona
Para Laura Serra, el factor de comportamiento es 0.90 y el factor de edad es 1.20.

La función multiplica los factores:

0.90 X 1.20 = 1.08

Por lo tanto, devuelve 1.08. Se trata de una puntuación calculada con la fórmula del ejercicio.

```python

def calcular_risc(factor_codi, factor_edat):
    ''' Recibe los dos factores obtenidos anteriormente, calcula su producto
    y devuelve el riesgo final.'''
    risc_final = factor_codi * factor_edat

    return risc_final
```

#### processar_clients()

Esta función recibe la lista completa de clientes. Recorre cada cliente y llama a las tres funciones anteriores para recuperar los datos de segmentación, obtener los factores y calcular el riesgo final.

Después guarda los datos originales y los resultados en el diccionario de diccionarios resultats.

La variable client representa un cliente individual en cada vuelta del bucle.

Por ejemplo, client["codi_comportament"] obtiene el código de comportamiento del cliente que se está procesando. Ese único valor se pasa a obtenir_segment_i_factor(), no la lista completa.

El resultado se utiliza junto con el factor de edad para calcular el riesgo. Después, se crea un diccionario con las diez columnas requeridas y se guarda en resultats, usando el identificador del cliente como clave.

Al terminar el bucle, resultats contiene a todos los clientes procesados.

```python

def processar_clients(clients):
    ''' Recibe la lista completa de clientes. Recorre cada cliente y llama a las tres funciones
    anteriores para recuperar los datos de segmentación, obtener los factores y calcular el riesgo final.
    Después guarda los datos originales y los resultados en el diccionario de diccionarios resultats.'''
    resultats = {}

    for client in clients:
        informacio = obtenir_segment_i_factor(client["codi_comportament"])

        factor_edat = obtenir_factor_edat(client["grup_edat"])

        risc_final = calcular_risc(informacio["factor"],factor_edat)

        id_client = client["id_client"]

        resultats[id_client] = {
            "id_cliente": client["id_client"],
            "nombre": client["nom"],
            "edad": client["edat"],
            "grupo_edad": client["grup_edat"],
            "codigo_comportamiento": client["codi_comportament"],
            "segmento": informacio["segmento"],
            "mensaje": informacio["mensaje"],
            "factor_código": informacio["factor"],
            "factor_edad": factor_edat,
            "riesgo_final": risc_final
        }

    return resultats
```

Esta función recibe el diccionario final y lo convierte en una tabla de pandas. Después ordena la tabla de mayor a menor riesgo final y la muestra por pantalla.

pd.DataFrame(resultats.values()) convierte los diccionarios de cada cliente en filas del DataFrame.

Finalmente, to_string(index=False) muestra la tabla sin una columna adicional de índices.

```python

def mostrar_dataframe(resultats):
    ''' Recibe el diccionario final y lo convierte en una tabla de pandas.
    Después ordena la tabla de mayor a menor riesgo final y la muestra por pantalla.'''

    dataframe = pd.DataFrame(resultats.values())

    dataframe = dataframe.sort_values (by="riesgo_final", ascending=False)

    print(dataframe.to_string(index=False))
```

4. Recuperación prioritaria: identifica al cliente con el riesgo más alto. Esto indica prioridad de contacto, pero no garantiza la mayor rentabilidad económica.
iloc[0] devuelve la primera fila de la tabla ordenada y iloc[-1] devuelve la última. Como la tabla está ordenada de mayor a menor, representan el riesgo máximo y el mínimo, respectivamente.


```python
import pandas as pd

def mostrar_respostes(resultats):
    ''' Recibe el diccionario final y lo convierte en una tabla de pandas.
    Después ordena la tabla de mayor a menor riesgo final y la muestra por pantalla.'''
    dataframe = pd.DataFrame(resultats.values())

    dataframe_ordenat = dataframe.sort_values(by="riesgo_final", ascending=False)

    # Pregunta 1: ¿A qué clientes contactar primero?
    print("\n1. Clientes prioritarios para retargeting:")

    print( dataframe_ordenat[
            ["nombre", "codigo_comportamiento", "riesgo_final"]].to_string(index=False))

    # Pregunta 2: mismo comportamiento, distinta edad
    print("\n2. Comparación de clientes con comportamiento 4:")

    clients_comportament_4 = dataframe[dataframe["codigo_comportamiento"] == 4]

    print( clients_comportament_4[
            ["nombre","grupo_edad","factor_código","factor_edad","riesgo_final"
            ]].to_string(index=False) )

    # Pregunta 3: ¿Quién muestra menor interés?
    print("\n3. Clientes con menor interés observable:")

    clients_interes_bajo = dataframe[dataframe["codigo_comportamiento"] == 0]

    print(clients_interes_bajo[ ["nombre", "segmento", "riesgo_final"]].to_string(index=False) )

    print("Estos clientes navegan sin añadir productos. "
    "Son candidatos de baja prioridad según los datos disponibles.")

    # Pregunta 4: ¿A quién priorizar para recuperar una compra?
    print("\n4. Cliente prioritario para intentar recuperar:")

    client_mayor_riesgo = dataframe_ordenat.iloc[0]

    print( client_mayor_riesgo["nombre"],"- Riesgo:",
        round(client_mayor_riesgo["riesgo_final"], 4))

    print( "El riesgo indica prioridad de contacto, pero no demuestra "
    "por sí solo qué recuperación será más rentable.")

    # Identificamos el riesgo mínimo
    client_menor_riesgo = dataframe_ordenat.iloc[-1]

    print("\nCliente con el riesgo final más bajo:")

    print(client_menor_riesgo["nombre"], "- Riesgo:",
     round(client_menor_riesgo["riesgo_final"], 4))
```

#### 1. Función convertir_a_diccionario()




Esta función se encarga de parsear la cadena de texto cruda, extraer los campos y estandarizar los síntomas eliminando espacios sobrantes. Devuelve un diccionario estructurado.

```python
def convertir_a_diccionario(linea_paciente):
    """
    Parsea una línea de texto cruda (formato: Nombre|Edad|Sintomas)
    y la convierte en un diccionario estructurado.
    """
    partes = linea_paciente.split("|")
    nombre = partes[0].strip()
    edad = partes[1].strip()
    sintomas_raw = partes[2].strip()

    return {"nombre": nombre, "edad": edad,"sintomas_raw": sintomas_raw}

```

#### 2. Función limpieza_datos()

Limpia la cadena de síntomas: pasa a minúsculas, elimina espacios en los extremos, divide por comas y reemplaza los espacios internos por guiones bajos.

```python
def limpieza_datos(sintomas_str):
    """
    Limpia la cadena de síntomas: pasa a minúsculas, elimina espacios en los extremos,
    divide por comas y reemplaza los espacios internos por guiones bajos.
    Devuelve una lista con los síntomas limpios.
    """
    lista_sintomas = sintomas_str.split(",")

    #  Creamos una lista vacía donde iremos guardando los síntomas ya limpios
    sintomas_limpios = []

    for sintoma in lista_sintomas:

        if sintoma.strip():

            texto_sin_extremos = sintoma.strip()

            texto_minusculas = texto_sin_extremos.lower()

            texto_final = texto_minusculas.replace(" ", "_")

            sintomas_limpios.append(texto_final)

    return sintomas_limpios
```

#### 3. Función asignar_riesgo()

Retorna la puntuación de riesgo de un síntoma individual usando el diccionario de referencia. Si el síntoma no está catalogado, se le asigna 0 por defecto.

```python
def asignar_riesgo_sintoma(sintoma, diccionario_riesgos):
    """
    Retorna la puntuación de riesgo de un síntoma individual usando el diccionario de referencia.
    Si el síntoma no está catalogado, se le asigna 0 por defecto.
    """
    return diccionario_riesgos.get(sintoma, 0)

```

#### Función calcular_riesgo()

Suma el riesgo de cada síntoma del paciente usando la fórmula: total = sum(sintomas). Creación de lista vacía para guardar las puntuaciones númericas de cada síntoma. Se utiliza bucle for para recorrer lista de síntomas uno por uno.
Llamada de función asignar_riesgo_sintomas(s, diccionario_riesgos), se le pasan los parámetros s, y la base de datos de riesgos. Se guarda la variable riego_de_este_sintoma a la lista puntuaciones. Se crea la variable: total_riesgo para guardar la adición sum(puntuaciones). Al final la función devuelve la variable: total_riesgo.

```python
def calcular_riesgo_total(sintomas_limpios, diccionario_riesgos):
    """
    Suma el riesgo de cada síntoma del paciente usando la fórmula: total = sum(sintomas)

    """
    puntuaciones = []


    for s in sintomas_limpios:

        riesgo_de_este_sintoma = asignar_riesgo_sintoma(s, diccionario_riesgos)


        puntuaciones.append(riesgo_de_este_sintoma)

    total_riesgo = sum(puntuaciones)

    return total_riesgo
```

#### 5. Función prioridad()
La función prioridad genera un informe estructurado y ordenado de mayor a menor riesgo utilizando pandas para facilitar la visualización del equipo médico.

```python
import pandas as pd

def priorizar_pacientes(pacientes_procesados):
    """
    Convierte la lista de pacientes procesados en un DataFrame de pandas,
    los ordena de mayor a menor prioridad y los muestra.
    """

    df = pd.DataFrame(pacientes_procesados)
    # Ordenar por el riesgo total de forma descendente
    df_ordenado = df.sort_values(by="Riesgo Total", ascending=False)
    return df_ordenado



```

### Función main_pacientes()
El main coordina todo el flujo de ejecución.

```python
def main_pacientes(lista_cruda, diccionario_riesgos):
    """
    Orquestador principal del proceso de triaje médico.
    """
    pacientes_procesados = []

    for linea in lista_cruda:

        paciente_dict = convertir_a_diccionario(linea)

        sintomas_limpios = limpieza_datos(paciente_dict["sintomas_raw"])

        riesgo_total = calcular_riesgo_total(sintomas_limpios, diccionario_riesgos)

        pacientes_procesados.append({
            "Paciente": paciente_dict["nombre"],
            "Edad": paciente_dict["edad"],
            "Síntomas Detectados": ", ".join(sintomas_limpios),
            "Riesgo Total": riesgo_total
        })

    reporte_prioridad = priorizar_pacientes(pacientes_procesados)

    print("=== INFORME DE PRIORIZACIÓN DE PACIENTES ===")
    display(reporte_prioridad)

```

```python
def procesar_seguros(clients, nivells_seguro):
    """
    Recorre la lista de clientes, busca los datos de su cobertura correspondientes
    en 'nivells_seguro', calcula el precio final utilizando la función especializada
    y devuelve una lista con los diccionarios de información completamente integrados.
    """
    clients_processats = []
    for cliente in clients:

        codigo = cliente["codi"]
        datos_seguro = nivells_seguro[codigo]

        precio_base = datos_seguro["preu_base"]
        precio_final = calcular_precio_final(precio_base, cliente["edat"])

        cliente_integrado = {
            "Nombre": cliente["nom"],
            "Edad": cliente["edat"],
            "Código": codigo,
            "Categoría": datos_seguro["categoria"],
            "Precio Base": precio_base,
            "Precio Final": precio_final,
            "Probabilidad Fraude": datos_seguro["fraude"]
        }

        clients_processats.append(cliente_integrado)

    return clients_processats
```

#### 3. Función generar_informe()
 Esta función recibe la lista de clientes procesados, la convierte en un DataFrame de pandas y ordena los clientes de mayor a menor según su probabilidad de fraude. Finalmente, devuelve el DataFrame ordenado para que pueda mostrarse en el informe.

```python
import pandas as pd
def generar_informe(clients_processats):
    """
    Convierte la lista de clientes procesados en un DataFrame de pandas,
    los ordena de mayor a menor riesgo y los muestra.
    """

    df = pd.DataFrame(clients_processats)
    df_ordenado = df.sort_values(by="Probabilidad Fraude", ascending=False)
    return df_ordenado
```

#### 4. Función analizar_resultados_seguros()
 Esta función analiza los resultados de los clientes para identificar quién tiene la mayor probabilidad de fraude, quién presenta el precio final más alto y quién debe tener prioridad en la revisión manual. Para seleccionar a este último, filtra los clientes de las categorías de riesgo alto y crítico, los ordena por probabilidad de fraude y, en caso de empate, por precio final. Finalmente, muestra los resultados del análisis por pantalla.

```python
def analizar_resultados_seguros(clients_processats):

    df = pd.DataFrame(clients_processats)

    idx_fraude = df['Probabilidad Fraude'].idxmax()
    cliente_fraude = df.loc[idx_fraude]

    idx_precio = df['Precio Final'].idxmax()
    cliente_precio = df.loc[idx_precio]

    clientes_revision = df[df["Categoría"].isin(["Risc alt", "Risc crític"])]

    cliente_prioritario = clientes_revision.sort_values(by=["Probabilidad Fraude", "Precio Final"],
    ascending=[False, False]).iloc[0]

    print("=== ANÁLISIS DE RESULTADOS ===")
    print(f"1. Cliente con más probabilidad de fraude: {cliente_fraude['Nombre']} ({cliente_fraude['Probabilidad Fraude'] * 100:.1f}% de probabilidad)")
    print(f"2. Cliente con el precio final más alto: {cliente_precio['Nombre']} ({cliente_precio['Precio Final']:.2f} €)")
    print(f"3. Cliente prioritario para revisar manualmente: {cliente_prioritario['Nombre']} (Categoría: {cliente_prioritario['Categoría']}, Probabilidad de Fraude: {cliente_prioritario['Probabilidad Fraude'] * 100:.1f}%)")

```

#### Función obtener_caracteres():
Recibirá el tipo de carácter que queremos obtener y devolverá
la lista correspondiente. Se crea un diccionario con las categorias, después se comprueba si el tipo de caracter existe en el diccionario por medio de bucle for para después convertir y devolver una lista con la cadena solicitada

```python
def obtener_caracteres(tipo):
    ''' Recibirá el tipo de carácter que queremos obtener y devolverá
    la lista correspondiente.'''

    categorias = {
        "mayusculas": "ABCDEFGHIJKLMNOPQRSTUVWXYZ",
        "minusculas": "abcdefghijklmnopqrstuvwxyz",
        "numeros": "0123456789",
        "signos": "!@#$%^&*()-_=+[]{};:,.?/"
    }

    if tipo in categorias:

        return list(categorias[tipo])
    else:
        raise ValueError("Tipo de carácter no válido")
```

#### Función validar_parametros(longitud,activos)
Esta función actúa como el primer filtro de seguridad, asegurando que se haya seleccionado al menos una categoría que este activa y que la longitud de la contraseña sea suficiente para incluir un carácter de cada tipo seleccionado.

```python
def validar_parametros(longitud, activos):
    '''
    Valida que los parámetros para la contraseña sean correctos.
    Comprueba que al menos una categoría esté activa y que la longitud
    sea suficiente para incluir al menos un carácter de cada tipo seleccionado.
    '''
    num_activos = 0
    for categoria in activos:
        if activos[categoria] == True:
            num_activos = num_activos + 1

    if num_activos == 0:
        print("Error de validación: Debe seleccionar al menos un tipo de carácter.")
        return False

    if longitud < num_activos:
        print(f"Error de validación: La longitud debe ser de al menos {num_activos} caracteres para la selección actual.")
        return False

    return True
```

#### Función  generar_base_obligatoria(activos)
Recibe el diccionario de categorías activas y devuelve una lista con un carácter aleatorio de cada una.
Se hace uso de la biblioteca numpy para seleccionar de manera aleatoria un carácter de cada categoría activa, garantizando así que no falte ningún tipo de carácter solicitado.

```python
import numpy as np

def generar_base_obligatoria(activos):
    '''
    Garantiza que la contraseña incluya al menos un carácter de cada tipo seleccionado.
    Recibe el diccionario de categorías activas y devuelve una lista con un carácter aleatorio de cada una.
    '''
    base = []
    for categoria, activo in activos.items():
        if activo:
            caracteres_disponibles = obtener_caracteres(categoria)
            caracter_elegido = np.random.choice(caracteres_disponibles)
            base.append(caracter_elegido)

    return base
```

#### Función completar_contraseña(base, longitud, pool_completo)
Rellena la longitud restante de la contraseña de manera uniforme seleccionando
caracteres aleatoriamente del pool_completo y añadiéndolos a la lista base.

```python
def completar_contraseña(base, longitud, pool_completo):
    '''
    Rellena la longitud restante de la contraseña de manera uniforme seleccionando
    caracteres aleatoriamente del pool_completo y añadiéndolos a la lista base.
    '''
    lista_resultado = list(base)

    diferencia = longitud - len(lista_resultado)

    if diferencia > 0:
        caracteres_restantes = np.random.choice(pool_completo, size=diferencia)
        lista_resultado.extend(caracteres_restantes)

    return lista_resultado
```

#### mezclar_contraseña(lista_resultado)
Recibe la lista completa con todos los caracteres seleccionados y desordena
aleatoriamente la posición de todos los elementos para eliminar la predictibilidad estructural. Se hace copia de la lista para no alterar la original directamente y después se usa numpy para mezclar la lista.

```python
def mezclar_contraseña(lista_resultado):
    '''
    Recibe la lista completa con todos los caracteres seleccionados (lista_resultado) y desordena
    aleatoriamente la posición de todos los elementos para eliminar la predictibilidad estructural.
    '''
    lista_mezclada = list(lista_resultado)

    np.random.shuffle(lista_mezclada)

    return lista_mezclada
```

#### Función crear_contraseña(...)
Orquesta el flujo completo de generación de contraseñas de forma segura. Primero se juntan los booleanos en un diccionario, después se validan los parámetros iniciales, luego se genera la base obligatoria de la contraseña. Para después consolidar la selección completa de caracteres permitidos. Para finalizar se completa la longitud requerida, mezclar aleatoriamente la lista de resultados para unir la lista y devolver la contraseña.

```python
def crear_contraseña(longitud, majuscules=True, minuscules=True, numeros=True, signes=False):
    '''
    Orquesta el flujo completo de generación de contraseñas de forma segura.
    '''
    activos = {
        "mayusculas": majuscules,
        "minusculas": minuscules,
        "numeros": numeros,
        "signos": signes
    }

    if not validar_parametros(longitud, activos):
        return None

    base = generar_base_obligatoria(activos)

    pool_completo = []
    for categoria, activo in activos.items():
        if activo:
            pool_completo.extend(obtener_caracteres(categoria))

    lista_completa = completar_contraseña(base, longitud, pool_completo)

    lista_mezclada = mezclar_contraseña(lista_completa)

    return  "".join(lista_mezclada)
```

#### Función leer_partidos()

Esta función se encarga de abrir el archivo historic partits.txt, leer todas las líneas y almacenar los partidos en una lista.
Elimina los saltos de línea y descarta los registros vacíos. Finalmente, devuelve la lista para que pueda ser procesada por las siguientes funciones.

```python

def leer_partidos(nombre_archivo):
    """
    Lee el archivo de texto y devuelve una lista
    con los partidos registrados.
    """
    with open(nombre_archivo, "r", encoding="utf-8") as archivo:

        lineas = archivo.readlines()

    partidos = []

    for linea in lineas:

        if linea.strip():
            partidos.append(linea.strip())

    return partidos
```

#### Función procesar_partidos()

Esta función es el núcleo del ejercicio. Recibe la lista de partidos, identifica a los dos equipos y sus goles, y acumula las estadísticas de cada uno en un diccionario de diccionarios. También determina quién gana, pierde o empata y actualiza los puntos correspondientes.

```python

def procesar_partidos(partidos):
    """
    Procesa todos los partidos y calcula las estadísticas
    de cada equipo: goles, victorias, empates, derrotas y puntos.
    """

    estadistiques = {}

    for partido in partidos:

        partes = partido.split("\t")

        equipo_local = partes[0]
        marcador = partes[1]
        equipo_visitante = partes[2]

        goles = marcador.split("-")

        goles_local = int(goles[0])
        goles_visitante = int(goles[1])

        for equipo in [equipo_local, equipo_visitante]:

            if equipo not in estadistiques:

                estadistiques[equipo] = {
                    "gols_favor": 0,
                    "gols_contra": 0,
                    "guanyats": 0,
                    "empatats": 0,
                    "perduts": 0,
                    "punts": 0
                }

        estadistiques[equipo_local]["gols_favor"] += goles_local
        estadistiques[equipo_local]["gols_contra"] += goles_visitante

        estadistiques[equipo_visitante]["gols_favor"] += goles_visitante
        estadistiques[equipo_visitante]["gols_contra"] += goles_local

        if goles_local > goles_visitante:

            estadistiques[equipo_local]["guanyats"] += 1
            estadistiques[equipo_local]["punts"] += 3

            estadistiques[equipo_visitante]["perduts"] += 1

        elif goles_visitante > goles_local:

            estadistiques[equipo_visitante]["guanyats"] += 1
            estadistiques[equipo_visitante]["punts"] += 3

            estadistiques[equipo_local]["perduts"] += 1

        else:

            estadistiques[equipo_local]["empatats"] += 1
            estadistiques[equipo_visitante]["empatats"] += 1

            estadistiques[equipo_local]["punts"] += 1
            estadistiques[equipo_visitante]["punts"] += 1

    return estadistiques
```

####  Función generar_dataframe()

Esta función recibe el diccionario de estadísticas de los equipos y lo convierte en un DataFrame de pandas. Organiza los nombres de los equipos en una columna llamada equip, conserva las seis estadísticas calculadas y ordena la clasificación de mayor a menor según los puntos.

```python

def generar_dataframe(estadistiques):
    """
    Convierte el diccionario de estadísticas en un DataFrame
    y ordena los equipos por puntos de mayor a menor.
    """

    dataframe = pd.DataFrame.from_dict(
        estadistiques,
        orient="index"
    )

    dataframe.index.name = "equip"

    dataframe = dataframe.reset_index()

    dataframe = dataframe[
        [
            "equip",
            "gols_favor",
            "gols_contra",
            "guanyats",
            "empatats",
            "perduts",
            "punts"
        ]
    ]

    dataframe_ordenat = dataframe.sort_values(
        by="punts",
        ascending=False,
        kind="stable"
    )

    return dataframe_ordenat.reset_index(drop=True)
```

#### Función mostrar_resultados()

Esta función recibe el DataFrame ordenado y analiza las estadísticas para identificar el equipo que ha marcado más goles y el que ha recibido más goles. Después muestra ambos resultados y presenta la clasificación final.

```python

def mostrar_resultados(dataframe):
    """
    Identifica los equipos con más goles a favor y en contra
    y muestra la clasificación final.
    """

    indice_goleador = dataframe["gols_favor"].idxmax()

    equipo_goleador = dataframe.loc[indice_goleador, "equip"]

    goles_marcados = dataframe.loc[
        indice_goleador, "gols_favor"
    ]

    indice_mas_goleado = dataframe["gols_contra"].idxmax()

    equipo_mas_goleado = dataframe.loc[
        indice_mas_goleado, "equip"
    ]

    goles_recibidos = dataframe.loc[
        indice_mas_goleado, "gols_contra"
    ]

    print("Equipo más goleador:", equipo_goleador)
    print("Goles marcados:", goles_marcados)

    print("\nEquipo más goleado:", equipo_mas_goleado)
    print("Goles recibidos:", goles_recibidos)

    print("\nCLASIFICACIÓN FINAL:")

    display(dataframe)
```

#### Función main_partidos()

Esta función es el coordinador principal del programa. Llama a las funciones anteriores para leer el archivo, procesar los partidos, construir el DataFrame y mostrar los resultados. Finalmente, devuelve el DataFrame de clasificación para poder reutilizarlo.

```python

def main_partidos(nombre_archivo):
    """
    Coordina el procesamiento del histórico de partidos,
    genera la clasificación y muestra los resultados.
    """

    partidos = leer_partidos(nombre_archivo)

    estadistiques = procesar_partidos(partidos)

    dataframe = generar_dataframe(estadistiques)

    mostrar_resultados(dataframe)

    return dataframe
```
